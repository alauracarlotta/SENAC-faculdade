```python
import pandas as pd
```

```python
df = pd.read_csv('world-happiness-report.csv', sep=',', on_bad_lines='skip')
df.head(5)
```

```python
df.info()
```

```python
# Média de Score de felicidade
mean_score = df['Happiness Score'].mean()
mean_score
```

```python
# Renomear a coluna GDP
df.rename(columns={'Economy (GDP per Capita)': 'PIB per Capita'}, inplace=True)

# Mediana do PIB per Capita
median_PIB = df['PIB per Capita'].median()
median_PIB
```

```python
# Desvio padrão no Score de felicidade
std_Score = df['Happiness Score'].std()
std_Score
```

```python
# Variância da expectativa de Vida Saudável
variance_healthy_life_expectancy = df['Health (Life Expectancy)'].var()
variance_healthy_life_expectancy
```
