```python
import pandas as pd
```

```python
df = pd.read_csv('books.csv', sep=',', on_bad_lines='skip')
```

```python
df.head() # Lê as 5 primeiras linhas do arquivo
```

```python
df.shape # Tamanho da base de dados [linhas x colunas]
```

```python
df.info() #cabecalhos de cada coluna - não há valores nulos [RangeIndex e Non-Null com mesmo valor]
```

```python
df[df.ratings_count == 0] # linha de dados com nenhuma avaliação - para o cenário, essas linhas não representam nenhuma ajuda. (total de 80 linhas)
```

```python
df.drop(df[df.ratings_count == 0].index, inplace=True)  # podemos remover as linhas com valor 0 diretamente do arquivo de banco de dados.
```

```python
df.shape # com o comando, confirmamos que aquelas linhas foram removidas da base de dados
```

```python
df.drop_duplicates() # verificar e remover linhas duplicadas
```

```python
# conforme valor acima, vemos que não temos nenhuma linha duplicada, já que a qtde de linhas são as mesmas (vide linha 14)
```

# Quais os 10 livros com mais avaliações?

```python
most_rated = df.sort_values('ratings_count', ascending=False)
most_rated.head(10)
```

```python
# para ter uma melhor visualização, apresentaremos com gráfico de barras, usando a biblioteca seaborn
```

```python
%pip install seaborn
```

```python
import seaborn as sns

sns.barplot(x=most_rated.head(10).ratings_count, y=most_rated.head(10).title)
```

# Quais as suas médias de avaliações?

```python
most_rated.head(10).average_rating.mean()
```
